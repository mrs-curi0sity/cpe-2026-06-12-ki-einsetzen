# Teil II: KI-Level: trainieren / finetunen oder API?

## Welche Methode für welches Problem?

| Problem | Methode | Level |
|---|---|---|
| Texte zusammenfassen, Briefe, Kodierung | Prompt / LLM | 0 – OpenWebUI |
| Suche in Leitlinien / PDFs | RAG | 0 – OpenWebUI |
| Risikoprognose aus Labordaten | Klassifikation | 1 – scikit-learn |
| Bildklassifikation (Histo, MRT) | CNN / ViT | 2–3 |
| Klinik-spezifisches Sprachmodell | Finetuning (LoRA) | 3 |

## Level 0: Kein Modell, kein Code

**Wann:** Einfache Automatisierung, Datentransformation, Berechnungen

**Beispiele**
- Excel-Formel / Power Query
- OpenWebUI: Dokumente hochladen → direkt fragen (RAG out of the box)
- OpenWebUI für Textentwürfe, Zusammenfassungen

**Aufwand:** Stunden | **Kosten:** 0 €

## Level 1: Python + pandas + scikit-learn

**Wann:** Tabellarische Daten, Klassifikation, Regression, Clustering

**Beispiele**
- Wiederaufnahme-Risiko aus Labordaten (XGBoost)
- Patientengruppen clustern (k-Means)
- Überlebensanalyse (Kaplan-Meier)

**Stack:** pandas, scikit-learn, matplotlib

**Aufwand:** Tage–Wochen | **Kosten:** ~0 €

## Level 2: Vortrainiertes Modell von HuggingFace

**Wann:** Bild- oder Textaufgaben, für die ein Modell bereits existiert

**Beispiele**
- Pathologie-Klassifikation mit vortrainiertem ResNet
- Named Entity Recognition in Arztbriefen
- Sentiment-Analyse von Patientenfeedback

**Stack:** transformers, torch, MONAI

**Aufwand:** Tage | **Kosten:** 0–50 €

## Level 3: Finetuning eines vortrainierten Modells

**Wann:** Domänenspezifische Sprache, spezifische Bildklassen

**Beispiele**
- LLM auf Klinik-Entlassbriefen finetunen (LoRA)
- Bildmodell auf eigenen MRT-Daten anpassen

**Stack:** transformers, PEFT/LoRA, unsloth

**Aufwand:** Wochen | **Kosten:** 500–5.000 €

## Level 4: Eigenes Modell from Scratch

**Wann:** Sehr spezifische Architektur, riesige proprietäre Datenmenge, Forschung

**Beispiele**
- Spezialisiertes Bildgebungsmodell für seltene Modalität
- Eigenes klinisches Sprachmodell (à la BioGPT)

**Stack:** PyTorch, CUDA, HPC-Cluster

**Aufwand:** Monate–Jahre | **Kosten:** 50.000–1.000.000+ €

## Übersicht: Was kostet welches Level?

| Level | Methode | Kosten | Aufwand |
|---|---|---|---|
| 0 | Excel / OpenWebUI | 0 € | Stunden |
| 1 | pandas + scikit-learn | 0 € | Tage–Wochen |
| 2 | Modell von HuggingFace | 0–50 € | Tage |
| 3 | Finetuning (LoRA) | 500–5.000 € | Wochen |
| 4 | Training from Scratch | 50.000–1 Mio. € | Monate–Jahre |

> Faustregel: Immer beim niedrigsten Level anfangen, das das Problem löst.

# Trends


## LLMs

KI bekommt eine Sprache.

**2017 — „Attention Is All You Need"** (Vaswani et al., Google): Der Transformer revolutioniert NLP und legt die Grundlage für alle modernen Sprachmodelle.

![](./images/trends_language.png){.slide-img}


## RAGs (Retrieval Augmented Generation)

KI kriegt eine Bibliothek.

**2020 — RAG-Paper** (Lewis et al., Meta): „Retrieval-Augmented Generation for Knowledge-Intensive NLP Tasks" — LLMs werden erstmals mit externer Wissensbasis kombiniert.

![](./images/trends_RAG.png){.slide-img}


## RLHF

Reinforcement Learning from Human Feedback. Vom Menschen lernen, wie ein Mensch zu reden.

**2022 — InstructGPT** (OpenAI): Erstmals wird RLHF großskalig eingesetzt, um Sprachmodelle auf menschliche Präferenzen auszurichten — Vorläufer von ChatGPT.

![](./images/trends_RLHF.png){.slide-img-sm}


## Generative AI

KI darf kreativ werden.

**2022 — Stable Diffusion / DALL·E 2** (Stability AI / OpenAI): Diffusionsmodelle ermöglichen hochqualitative Bildgenerierung aus Text — Generative AI wird Mainstream.

![](./images/generative_AI.png){.slide-img}


## Knowledge Graphen + LLMs

Wissenslandkarten werden aufgebaut.

**2023 — GraphRAG** (Microsoft): Strukturiertes Wissen aus Knowledge Graphs wird mit LLMs verknüpft — präzisere, nachvollziehbarere Antworten durch semantische Vernetzung.

![](./images/trends_knowledge_graph.png){.slide-img}


## Multimodalität

KI versteht nicht mehr nur Text — sondern Bilder, Sprache, Video, Dokumente gleichzeitig.

**2023 — GPT-4V / Gemini** (OpenAI / Google): Multimodale Großmodelle verarbeiten Text und Bilder in einem einzigen Modell — Startschuss für breite multimodale Anwendungen.

![](./images/trends_multimodal.png){.slide-img-sm}


## Agentic Systems

KI bekommt Werkzeuge.

**2023 — Toolformer / Function Calling** (Meta / OpenAI): Modelle lernen, externe Tools (APIs, Code, Suche) eigenständig aufzurufen — Grundlage für autonome KI-Agenten.

![](./images/trends_tools.png){.slide-img}


## Reasoning-Modelle

Neue Modelle "denken nach", bevor sie antworten — statt sofort zu generieren.

**2024 — OpenAI o1**: Chain-of-Thought-Reasoning wird intern skaliert — Modelle lösen komplexe Logik- und Matheaufgaben durch explizites Nachdenken vor der Antwort.

![](./images/trends_reasoning.png){.slide-img}


## Robotics

KI bekommt einen Körper und sensorische (eigene!) Erfahrung.

**2024 — Figure 01 / RT-2** (Figure AI / Google DeepMind): LLM-gesteuerte Roboter agieren in der realen Welt — KI-Reasoning trifft auf physische Embodiment.

![](./images/trends_sensoric_input.png){.slide-img}


## Vibe Coding

KI wird zum Programmierer — auch ohne Programmierkenntnisse.

**2023–2025 — GitHub Copilot → Cursor → Claude Code / Lovable**: Natürlichsprachliche Prompts erzeugen lauffähigen Code. Softwareentwicklung öffnet sich für alle.

![](./images/trends_vibe_coding.png){.slide-img}
