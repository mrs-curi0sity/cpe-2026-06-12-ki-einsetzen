# Teil II: KI-Level: trainieren / finetunen oder API?

## Level 0: Kein Code, kein Modell

- **Was?** Einfache Automatisierung, Datentransformation, Berechnungen
- **Wann?** Keine KI nötig – klassische Tools reichen
- **Kosten:** 0 €
- **Datenschutz:** ✅ Lokal, volle Kontrolle
- **Beispiele:** Excel, PowerQuery

![](./images/level_00_cartoon_excel.png){.absolute bottom=20 right=20 width="220px"}

## Level 1: Klassisches ML (Python + scikit-learn)

- **Was?** Klassifikation, Regression, Clustering
- **Wann?** Strukturierte Daten, kein LLM nötig
- **Kosten:** ~0 €
- **Datenschutz:** ✅ Lokal
- **Beispiele:** Kleine Vorhersagemodelle (Regression), 
  Patientengruppen clustern (k-Means)

![](./images/level_01_cartoon_python.png){.absolute bottom=20 right=20 width="220px"}

## Level 2: API (OpenAI, Anthropic, ...)

- **Was?** Fertiges LLM in der Cloud anfragen, kein Setup nötig
- **Wann?** Schneller Start, Prototyping, fehlende Entwicklerressourcen
- **Kosten:** ~0,001–0,01 € pro 1.000 Tokens / „ein paar Euro pro Monat und User"
- **Datenschutz:** ⚠️ nicht gegeben! ⚠️
- **Beispiel:** ChatGPT API

![](./images/level_02_cartoon_api.png){.absolute bottom=20 right=20 width="220px"}

## Level 3: Lokales Modell (+ RAG)

- **Was?** Fertiges LLM z.B. von [Huggingface](https://huggingface.co/models) lokal installieren und betreiben
- **Wann?** Datenschutz kritisch, kein Cloud-Zugang gewünscht
- **Kosten:** ~1.000 € (GPU)
- **Datenschutz:** ✅ Daten bleiben intern
- **Beispiel:** Llama 3 lokal

![](./images/level_03_cartoon_local.png){.absolute bottom=20 right=20 width="220px"}

## Level 4: Finetuning

- **Was?** Bestehendes Modell auf eigene Daten weiter trainieren
- **Wann?** Spezialfacette eines bestehenden Problems
- **Kosten:** ~10.000 €
- **Datenschutz:** ✅ Kontrollierbar
- **Beispiel:** GPT für medizinische Fachsprache

![](./images/level_04_cartoon_finetune.png){.absolute bottom=20 right=20 width="220px"}

## Level 5: Training from Scratch

- **Was?** Komplett selbst trainieren – von Grund auf
- **Wann?** Sehr spezifisches Problem, große Datenmengen, großes Budget
- **Kosten:** ~100.000 €
- **Datenschutz:** ✅ Lokal, volle Kontrolle
- **Beispiel:** AlphaFold (Proteinfaltung)

![](./images/level_05_cartoon_toddler.png){.absolute bottom=20 right=20 width="220px"}

## Zusammenfassung

| L | Was | Kosten | DS | Aufwand |
|--|--------|----------|---|---------|
| 0 | Kein Code | 0 € | ✅ | S |
| 1 | Klassisches ML | 0 € | ✅ | M |
| 2 | API | 💰 (variabel) | ⚠️ | S |
| 3 | Lokales Modell | 💰 | ✅ | S |
| 4 | Finetuning | 💰💰 | ✅ | M |
| 5 | Training from Scratch | 💰💰💰 | ✅ | L |

# Teil III: Tools

## Tabellarische Daten (Julius.ai)

**📊 Datenanalyse per Chatbot**

[julius.ai](https://julius.ai) · kostenlos · kein Account nötig

- Excel oder CSV hochladen, dann auf Deutsch Fragen stellen
- Erstellt Grafiken, berechnet Statistiken, erklärt Ergebnisse
- Ideal für Auswertungen ohne SPSS- oder R-Kenntnisse

**Oder einfach: Claude, ChatGPT**


## Bildbearbeitung (Cleanup Pictures)

**✂️ Hintergründe & Objekte aus Fotos entfernen**

[cleanup.pictures](https://cleanup.pictures) · kostenlos · kein Account nötig

- Bild hochladen, unerwünschte Bereiche markieren → weg
- Funktioniert mit Hintergründen, Personen, Objekten, Wasserzeichen
- Ideal für Patientenfotos, Wundbilder, Lehrbilder professionell freistellen


## Videogenerierung (Kling oder Midjourney)

**🎬 Videos aus Text oder Bildern erstellen**

[klingai.com](https://klingai.com) · nur wenige Versuche kostenlos · Account nötig

- Text oder Bild eingeben, fertiges Video erscheint in wenigen Sekunden
- Erstellt realistische Bewegtbilder, auch aus eigenen Fotos
- Ideal zum Zeigen, dass KI nicht mehr nur Text und Bilder kann

## Bildgenerierung: Midjourney

**🖼️ Beste KI-Bildgenerierung aktuell**

[midjourney.com](https://midjourney.com) · **$10/Monat** · Account erforderlich

- Text-Prompt eingeben → professionelle, künstlerische Bilder in Sekunden
- Qualitativ weit vor kostenlosen Alternativen
- Ideal für Lehrbilder, Broschüren, Präsentationen, Patientenmaterialien

> *"Freundliche Pflegeperson erklärt einer älteren Patientin etwas, 
> helles Krankenhauszimmer, realistisch"*
>


## Musikgenerierung: Suno

**🎵 Eigene Songs per Texteingabe erstellen**

[suno.com](https://suno.com) · kostenlos · kein Account nötig

- Kurzen Text oder Stimmung eingeben, fertiger Song erscheint in Sekunden
- Wählt automatisch Genre, Instrumente und Gesang
- Ideal zum Zeigen, wie weit KI-Kreativität schon reicht

## Sprachtranskription: Whisper

**🎙️ Sprache zu Text per KI**

[huggingface.co/spaces/openai/whisper](https://huggingface.co/spaces/openai/whisper) · kostenlos · kein Account nötig

- Audiodatei hochladen oder Mikrofon nutzen, Text erscheint sofort
- Funktioniert auf Deutsch mit hoher Genauigkeit, über 99 Sprachen
- Open Source von OpenAI – auch per API oder lokal nutzbar

## Übersetzung (DeepL)

**🌍 KI-Übersetzung in Profi-Qualität**

[deepl.com](https://deepl.com) · kostenlos · kein Account nötig

- Übersetzt Text und ganze Dokumente (Word, PDF)
- besser als Google Translate, besonders bei Fachsprache
- Ideal für internationale Studien, Arztbriefe für fremdsprachige Patient:innen

> *Englische Leitlinie hochladen → auf Deutsch lesen und direkt weiterarbeiten*

## Präsentationen erstellen (Gamma.app)

**📑 KI erstellt Präsentationen automatisch**

[gamma.app](https://gamma.app) · kostenlos* · Account erforderlich

- Text oder Stichpunkte eingeben → fertige Präsentation in Sekunden
- Ca. 10 Präsentationen kostenlos, danach kostenpflichtig
- Ideal für Fortbildungen, Patientenaufklärung, interne Berichte

> *"Erstelle eine Patientenbroschüre zum Thema Wundversorgung."*

## Recherche mit Quellen (Perplexity AI)

**🔍 KI-Suche mit Quellenangaben**

[perplexity.ai](https://perplexity.ai) · kostenlos · kein Account nötig

- Stellt Fragen in natürlicher Sprache statt Keywords
- Jede Antwort kommt mit verlinkten Quellen
- Ideal für schnelle Literaturrecherche ohne PubMed-Kenntnisse

> *"Welche Evidenz gibt es für intermittierendes Fasten bei Typ-2-Diabetes?"*

## Recherche mit wissenschaftlichen Quellen (Consensus)

[consensus.app](https://consensus.app) · kostenlos* · Account empfohlen

- Nur Peer-Review-Literatur (PubMed, Semantic Scholar, OpenAlex)
- Zeigt direkt die Mehrheitsmeinung der Studien
- Bewertet nach Zitationszahl & Journal-Reputation
- Stärke: MINT & Medizin · Schwäche: Geisteswissenschaften
- Keine halluzinierten Quellen

> *"Reduziert intermittierendes Fasten das Risiko für kardiovaskuläre Erkrankungen?"*

## Dokumente verstehen (NotebookLM)

**📓 KI-Assistent für eigene Dokumente**

[notebooklm.google.com](https://notebooklm.google.com) · kostenlos · Google-Account nötig

- PDF, Word oder Text hochladen, dann Fragen zum Inhalt stellen
- Erstellt automatisch Zusammenfassungen, FAQs und Mindmaps
- Highlight: generiert einen fertigen Podcast-Dialog aus dem Dokument

# Teil IV: Trends


## LLMs

KI bekommt eine Sprache.

**2017 — „Attention Is All You Need"** (Vaswani et al., Google): Der Transformer revolutioniert NLP und legt die Grundlage für alle modernen Sprachmodelle.

![](./images/trends_language.png){.slide-img}


## RAGs (Retrieval Augmented Generation)

KI kriegt eine Bibliothek.

**2020 — RAG-Paper** (Lewis et al., Meta): „Retrieval-Augmented Generation for Knowledge-Intensive NLP Tasks" — LLMs werden erstmals mit externer Wissensbasis kombiniert.

![](./images/trends_RAG.png){.slide-img}


## RLHF

Reinforcement Learning from Human Feedback. Vom Menschen lernen, wie ein Mensch zu reden.

**2022 — InstructGPT** (OpenAI): Erstmals wird RLHF großskalig eingesetzt, um Sprachmodelle auf menschliche Präferenzen auszurichten — Vorläufer von ChatGPT.

![](./images/trends_RLHF.png){.slide-img-sm}


## Generative AI

KI darf kreativ werden.

**2022 — Stable Diffusion / DALL·E 2** (Stability AI / OpenAI): Diffusionsmodelle ermöglichen hochqualitative Bildgenerierung aus Text — Generative AI wird Mainstream.

![](./images/trends_generative_ai.png){.slide-img}


## Knowledge Graphen + LLMs

Wissenslandkarten werden aufgebaut.

**2023 — GraphRAG** (Microsoft): Strukturiertes Wissen aus Knowledge Graphs wird mit LLMs verknüpft — präzisere, nachvollziehbarere Antworten durch semantische Vernetzung.

![](./images/trends_knowledge_graph.png){.slide-img}


## Multimodalität

KI versteht nicht mehr nur Text — sondern Bilder, Sprache, Video, Dokumente gleichzeitig.

**2023 — GPT-4V / Gemini** (OpenAI / Google): Multimodale Großmodelle verarbeiten Text und Bilder in einem einzigen Modell — Startschuss für breite multimodale Anwendungen.

![](./images/trends_multimodal.png){.slide-img-sm}


## Agentic Systems

KI bekommt Werkzeuge.

**2023 — Toolformer / Function Calling** (Meta / OpenAI): Modelle lernen, externe Tools (APIs, Code, Suche) eigenständig aufzurufen — Grundlage für autonome KI-Agenten.

![](./images/trends_tools.png){.slide-img}


## Reasoning-Modelle

Neue Modelle "denken nach", bevor sie antworten — statt sofort zu generieren.

**2024 — OpenAI o1**: Chain-of-Thought-Reasoning wird intern skaliert — Modelle lösen komplexe Logik- und Matheaufgaben durch explizites Nachdenken vor der Antwort.

![](./images/trends_reasoning.png){.slide-img}

## Robotics

KI bekommt einen Körper.

**2024 — Figure 01 / RT-2 / Atlas electric** (Figure AI / Google DeepMind / Boston Dynamics):
LLMs steuern erstmals physische Roboter in der realen Welt — Sprache wird zur Handlungsanweisung.

![](./images/trends_robotics_figure01.png){.slide-img}


## Vibe Coding

KI wird zum Programmierer — auch ohne Programmierkenntnisse.

**2023–2025 — GitHub Copilot → Cursor → Claude Code**: Natürlichsprachliche Prompts erzeugen lauffähigen Code. Softwareentwicklung öffnet sich für alle.

![](./images/trends_vibe_coding.png){.slide-img}
